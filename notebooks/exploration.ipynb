# DprE1 Benchmark: Exploratory Analysis & Model Comparison

This notebook demonstrates how to load the master benchmark dataset, evaluate predictive metrics, and explore the discordance between ligand-based ML and structure-based AutoDock Vina docking.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_auc_score, average_precision_score

# Load master predictions
df = pd.read_csv('../data/processed/master_predictions_phase5.csv')
print(f"Loaded {len(df)} non-covalent molecules.")
df.head()
```

```python
# Summary of ROC-AUC across models
y_true = (df['label'] == 'Active').astype(int).values
print("Track B (Cluster CV):")
print(f"  Random Forest ROC-AUC:      {roc_auc_score(y_true, df['prob_rf_cluster']):.3f}")
print(f"  Logistic Regression ROC-AUC: {roc_auc_score(y_true, df['prob_lr_cluster']):.3f}")
print(f"  AutoDock Vina ROC-AUC:       {roc_auc_score(y_true, df['vina_score']):.3f}")
print(f"  RRF Hybrid ROC-AUC:          {roc_auc_score(y_true, df['rrf_rf_vina_cluster']):.3f}")

print("\nTrack C (Leave-Hydantoin-Out):")
print(f"  Random Forest ROC-AUC:      {roc_auc_score(y_true, df['prob_rf_lho']):.3f}")
print(f"  Logistic Regression ROC-AUC: {roc_auc_score(y_true, df['prob_lr_lho']):.3f}")
print(f"  AutoDock Vina ROC-AUC:       {roc_auc_score(y_true, df['vina_score']):.3f}")
print(f"  RRF Hybrid ROC-AUC:          {roc_auc_score(y_true, df['rrf_rf_vina_lho']):.3f}")
```
